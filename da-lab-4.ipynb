```python
# importing necessary library
import pandas as pd
# load the dataset
df = pd.read_csv(r'/kaggle/input/udemy-courses-dataset/UdemyCoursesDataset.csv')
```

```python
# Display the first 6 rows
df.head(6)
```

### **1. What are the list of the subjects for which Udemy is offering courses?**

```python
subjects = df['subject'].unique()
print(subjects)
```

### **2. Which subject has the maximum number of courses?**

```python
max_subject = df['subject'].value_counts().idxmax()
print(max_subject)
```

### **3. What is the number of the courses which are Free of Cost?**

```python
free_courses = df[df['is_paid']== False].shape[0]
print(f"Number of free courses: {free_courses}")
```

### **4. Show all the courses which are Paid?**

```python
paid_courses = df[df['is_paid']== True].shape[0]
print(f"Number of paid courses: {paid_courses}")
```

### **5. Which are Top Selling Courses?**

```python
top_selling = df.sort_values(by='num_subscribers', ascending=False)
top_selling[['course_title', 'num_subscribers']].head()
```

### **6. How many number of courses are there in the Intermediate Level?**

```python
intermediate_count = df[df['level'] == 'Intermediate Level'].shape[0]
print(f"Number of Intermediate Level Courses: {intermediate_count}")
```

### **7. Show all courses of Graphic Design where the price is below 100?**

```python
graphic_design_courses = df[(df['subject'] == 'Graphic Design') & (df['price'] <= '100')]
graphic_design_courses
```

### **8. List out all the courses that are related with `Python`?**

```python
python_courses = df[df['course_title'].str.contains('python', case=False)]
python_courses
```

### **9. What are courses that published in year 2015 ?**

```python
df['published_timestamp'] = pd.to_datetime(df['published_timestamp'])

courses_2015 = df[df['published_timestamp'].dt.year == 2015]
courses_2015
```
